### Remember

| Idea | Meaning |
|------|---------|
| Model | $f_{w,b}(x) = wx + b$ |
| Parameters | $w$, $b$ |
| Cost $J(w,b)$ | score of how wrong $w,b$ are |
| Minimize $J$ | find $w,b$ with the **lowest** cost |

$$
J(w,b) = \frac{1}{2m} \sum_{i=0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2
$$
