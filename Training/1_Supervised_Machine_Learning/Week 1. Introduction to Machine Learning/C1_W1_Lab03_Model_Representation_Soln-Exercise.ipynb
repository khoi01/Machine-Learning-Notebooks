## Exercise 5 — Write `compute_model_output`

This function applies the same formula to **every** example:

$$
f_{w,b}(x^{(i)}) = w x^{(i)} + b
$$

**Task:** complete the loop.  
Hint from Sample: `f_wb[i] = w * x[i] + b`

```python
def compute_model_output(x, w, b):
    """
    Predict for every value in x using f = w*x + b.
    """
    m = x.shape[0]
    f_wb = np.zeros(m)

    for i in range(m):
        # TODO: apply the linear regression formula
        f_wb[i] = w * x[i] + b

    return f_wb
```

```python
# CHECK — Exercise 5
preds = compute_model_output(x_train, w=200, b=50)
print(f"predictions = {preds}")
assert np.allclose(preds, [250.0, 450.0, 650.0]), (
    'With w=200, b=50, predictions should be [250, 450, 650]'
)
print('Exercise 5 passed!')
```

**Task:**

```python
w = 200
b = 50

# TODO: get predictions
predictions = compute_model_output(x_train,w,b)

# TODO: plot line and scatter
plt.plot(x_train,predictions,c='b',label='Our Prediction')
plt.scatter(x_train,y_train,marker='x',c='r',label="Actual values")

plt.title('Cafe Sales — Model vs Data')
plt.xlabel('Customers (hundreds)')
plt.ylabel('Sales (in $100s)')
plt.legend()
plt.show()

print(f"predictions = {predictions}")
print(f"actual y    = {y_train}")
```

## Exercise 8 — What if $w$ and $b$ are wrong?

**Task:** try a bad model with `w=50`, `b=0`.

Compare predictions to actual `y_train`.  
You should see they do **not** match — that is why later labs learn better $w$ and $b$.

```python
# TODO: set bad parameters and predict
w_bad = 50
b_bad = 0
pred_bad = compute_model_output(x_train,w_bad,b_bad)

plt.plot(x_train, pred_bad, c='b', label='Bad prediction')
plt.scatter(x_train, y_train, marker='x', c='r', label='Actual values')
plt.title('Bad w and b')
plt.xlabel('Customers (hundreds)')
plt.ylabel('Sales (in $100s)')
plt.legend()
plt.show()

print(f"bad predictions = {pred_bad}")
print(f"actual y        = {y_train}")
```

```python
# CHECK — Exercise 8
assert w_bad == 50 and b_bad == 0, 'Use w_bad=50 and b_bad=0'
assert np.allclose(pred_bad, [50.0, 100.0, 150.0]), 'Predictions should be [50, 100, 150]'
assert not np.allclose(pred_bad, y_train), 'Bad model should NOT match y_train'
print('Exercise 8 passed! You can see why good w,b matter.')
```
